```python
dbutils.widgets.text("table_name","","TABLE_NAME")

table_name = dbutils.widgets.get("table_name")
```

```python

read_base_path = f"file:/Workspace/Shared/Retail_Project/Raw_Data/{read_container_name}"
```

```python
write_base_path = f"file:/Workspace/Shared/Retail_Project/Raw_Data/{write_container_name}"
```

```python
from datetime import datetime
current_date = datetime.now().strftime("%Y%m%d")
```

```python
df1 = spark.read.csv(f"{read_base_path}/{table_name}",header=True,inferSchema=True)
df1.display()
```

```python
from pyspark.sql.functions import current_timestamp
df1 = df1.withColumn("ingest_ts",current_timestamp())
```

```python
df1.display()
```

```python
df1.coalesce(1).write.format("csv").mode("overwrite").save(
    f"{write_base_path}/{table_name}"
)
```
